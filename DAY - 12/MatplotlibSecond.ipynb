```python
products=['Laptop','Ipad','Headphone','Keyboard']
sales=[400,200,600,150]
plt.figure(figsize=(7,7))
plt.pie(sales, labels=products,autopct='%1.1f%%',   # shows percentages with 1 decimal place
        startangle=90)   # optional: rotate chart for better look

plt.title('Products Weekly Sales Report')
plt.show()
```

# Scatter Plots
# In scatter plots we represent data in points, where each points represent a data observation. We can check how to values correlate

```python
product_sales={'Laptop':750,'Smartphone':650,'Smartwatch':400,'Tablets':500,'AC':400}
products=list(product_sales.keys())
sales=list(product_sales.values())
plt.scatter(products,sales)
plt.title('Basic product and sales represent')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.grid('True')
plt.show()
```

```python
players=['MSD','ROHIT','VIRAT','GILL','SHAMI']
match1_score=[45,67,56,78,89]
match2_score=[55,78,40,80,97]
match3_score=[90,40,20,30,60]
plt.figure(figsize=(8,5))
plt.scatter(players,match1_score,color='red',label='Match One',s=100)
plt.scatter(players,match2_score,color='green',label='Match Two',s=100,marker='d')
plt.scatter(players,match3_score,color='blue',label='Match Three',s=150,marker='p')
plt.title('Comparision of Player in Two Matches')
plt.xlabel('Players')
plt.ylabel('Score')
plt.legend(loc='best')  # best(dia akan masuk kan mana paling sesuai...kalau nak pilih sendiri kena masukkan upper left,lower left, left, center, upper center ...and so on right ,right left...
plt.grid(True,linestyle='--',alpha=0.5)
plt.show()
```

# Write a comparision that shows average calories burned for each workout type from final_data dataset

```python
df=pd.read_csv('Final_data.csv')
avg_cal=df.groupby('Workout_Type')['Calories_Burned'].mean()
selected_index=avg_cal.index
selected_value=avg_cal.values
plt.scatter(selected_index,selected_value,color='green',s=100,marker='p')
plt.title('Comparision of Workout and Calories Burned')
plt.xlabel('Workout Type')
plt.ylabel('Calories Burned')
plt.grid(True,linestyle='-',alpha=0.5)
plt.show()
```

# Create a bar chart using following list data

    ## python_lib=['Numpy','Pandas','Matplotlib','Scikit-learn']
# Liked by users as follows
    ## Liked_by=[5000,7000,9000,4000]
