# First example to create a line chart from our data
    ## Students.xlsx for fetching the data
    ## Generate line chart based on Student Name and Student Marks

```python
import pandas as pd
```

```python
df=pd.read_excel('C:/Users/User/OneDrive/Desktop/AIML/DAY - 10/students.xlsx')
```

```python
names=df['Name']
marks=df['Marks']
```

```python
plt.plot(names,marks,marker='s',color='green')
plt.title('Students Marks Line Chart')
plt.xlabel('Student Name')
plt.ylabel('Student Marks')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(5,3))  # Size chart (panjang & lebar chart tu besar or kecik mana kita nak)
plt.plot(names,marks,marker='o',color='blue')
plt.title('Students Progress Chart')
plt.xlabel('---Name---')
plt.ylabel('---Marks---')
plt.grid(True)
plt.show()
```

```python
print('Solid Line')
plt.figure(figsize=(5,3))
plt.plot(names,marks,marker='o',color='blue',linestyle='-')  # linestyle bentuk line tu lah, boleh pilih
plt.title('Students Progress Chart')
plt.xlabel('---Name---')
plt.ylabel('---Marks---')
plt.grid(True)
plt.show()
```

```python
print('Dashed Line')
plt.plot(names,marks,marker='o',color='blue',linestyle='--')
plt.show()
```

```python
print('Dotted Line')
plt.plot(names,marks,marker='o',color='blue',linestyle=':')
plt.show()
```

```python
print('Dashed Dotted Line')
plt.plot(names,marks,marker='o',color='blue',linestyle='-.')
plt.show()
```

```python
plt.figure(figsize=(5,3))
plt.plot(names,marks,marker='v',color='blue',linestyle='-.') 
plt.title('Students Progress Report')
plt.xlabel('---Name---')
plt.ylabel('---Marks---')
plt.grid(True)
plt.show()
```
