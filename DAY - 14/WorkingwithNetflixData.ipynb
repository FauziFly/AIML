```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv(r'C:\Users\User\OneDrive\Desktop\AIML\DAY - 13\netflix_data.csv')
df.head(3)
```

```python
sns.set_theme(style='whitegrid',palette='dark')
sns.countplot(data=df,x='type')
plt.title('Number of movies vs number of tv show')
plt.xlabel('Content type')
plt.ylabel('Count')
plt.show()
```

```python
sns.set_theme(style='darkgrid',palette='Set2')
sns.countplot(data=df,x='type')
plt.title('Number of movies vs number of tv show')
plt.xlabel('Content type')
plt.ylabel('Count')
plt.show()
```

```python
df=df.dropna(subset=['type','genre'])
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df.head(10),x='type',hue='country',palette='Set2')
plt.title('Netflix Content Type by Country')
plt.xlabel('Content type')
plt.ylabel('Count')
plt.legend(title='Country')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df.head(50),x='type',hue='country',palette='Set2')
plt.title('Netflix Content Type by Country only top 10')
plt.xlabel('Content type')
plt.ylabel('Count')
plt.legend(title='Country',ncol = 5,bbox_to_anchor=(1.02,1),loc='upper left',frameon=True)
plt.show()
```

# Exercise:
## Use Netflix dataset to visualize the distribution of release years for Movies Vs TV Shows using Seaborn's histplot()
## Use different color for Movie Vs TV Show

```python
sns.set_theme(style='whitegrid')
plt.figure(figsize=(8,5))
sns.histplot(data=df, x='release_year', hue='type', multiple='stack', palette='Set2', bins=30)
plt.title('Distribution of Release Years for Movies vs TV Shows', fontsize=13)
plt.xlabel('Release Year')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df,x='release_year',hue='type',multiple='dodge',palette='Set2',bins=30)
plt.title('Side-by-Side Distribution of Release Years')
plt.xlabel('Release Year')
plt.ylabel('Count')
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
```
