# Seaborn Library pf Matplotlib
## Seaborn is a data visualization library in Matplotlib that provides high level interface for creating interactive and more colorful Plots with less code.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
palette=sns.color_palette()
print(palette)
```

```python
sns.palplot(palette)
```

```python
palette_green=sns.color_palette('Greens',20)
sns.palplot(palette_green)
palette_red=sns.color_palette('Reds',20)
sns.palplot(palette_red)
palette_orange=sns.color_palette('Oranges',20)
sns.palplot(palette_orange)
```

```python
data_dict={'building':['burj-khalifa','medrika','effil-tower','Qutub minar'],
           'category':['height','medium','low','medium'],
           'building_height':[150,120,100,110]}

bar_colors=['red','green','purple']
data=pd.DataFrame(data_dict)
sns.barplot(x='building',y='building_height',data=data,hue='category',palette=bar_colors)
plt.show()
```

```python
data_iris=sns.load_dataset('iris')
data_iris
#data_flight=sns.load_dataset('flights')
#data_flight
```

```python
sns.barplot(x='species',y='sepal_width',data=data_iris, hue='species',palette=bar_colors)
plt.title('IRIS SPECIES DISTRIBUTION BAR CHART')
plt.xlabel('NAME OF SPECIES')
plt.ylabel('SEPAL WIDTH')
plt.grid(axis='y',linestyle='--',color='lightblue',alpha=0.5)
plt.show()
plt.tight_layout()
```

```python
sns.violinplot(x='species',y='sepal_width',data=data_iris,hue='species',palette='pastel')
plt.title('IRIS SPECIES DISTRIBUTION BAR CHART')
plt.xlabel('NAME OF SPECIES')
plt.ylabel('SEPAL WIDTH')
plt.grid(axis='y',linestyle='--',color='lightblue',alpha=0.5)
plt.show()
plt.tight_layout()
```

```python
sns.stripplot(x='species',y='sepal_width',data=data_iris,hue='species',palette='pastel')
plt.title('IRIS SPECIES DISTRIBUTION BAR CHART')
plt.xlabel('NAME OF SPECIES')
plt.ylabel('SEPAL WIDTH')
plt.grid(axis='y',linestyle='--',color='lightblue',alpha=0.5)
plt.show()
plt.tight_layout()
```

```python
sns.pairplot(data=data_iris,hue='species')
plt.show()
```

```python
sns.set_style('darkgrid')
sns.scatterplot(x='sepal_length',y='sepal_width',hue='species',data=data_iris)
plt.title('Darkgrid Theme Example using Iris Data')
plt.show()
```

```python
sns.set_style('whitegrid')
sns.scatterplot(x='sepal_length',y='sepal_width',hue='species',data=data_iris)
plt.title('Darkgrid Theme Example using Iris Data')
plt.show()
```

```python
sns.set_style('white')
sns.scatterplot(x='sepal_length',y='sepal_width',hue='species',data=data_iris)
plt.title('Darkgrid Theme Example using Iris Data')
plt.show()
```

```python
sns.set_style('dark')
sns.scatterplot(x='sepal_length',y='sepal_width',hue='species',data=data_iris)
plt.title('Darkgrid Theme Example using Iris Data')
plt.show()
```

```python
sns.set_style('ticks')
sns.scatterplot(x='sepal_length',y='sepal_width',hue='species',data=data_iris)
plt.title('Darkgrid Theme Example using Iris Data')
plt.xlabel('Sepal length (cm)')
plt.ylabel('Sepal width (cm)')
plt.xticks([4,5,6,7,8], fontsize=10, color='darkblue')
plt.yticks([1,2,3,4,5,6,7,], fontsize=10, color='purple')
sns.despine()
plt.show()
```
