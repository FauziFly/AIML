```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.random.rand(50)
y=np.random.rand(50)
temp=np.random.randint(20,100,size=50)

plt.scatter(x,y,c=temp,cmap='plasma', s=120, edgecolor='black')
plt.colorbar(label='Temperature')
plt.title('Scatter Plot Colored by Temperature')
plt.xlabel('X-Random')
plt.ylabel('Y-Random')
plt.show()
```

```python
height_cm=[150,155,160,165,170,175,180,185]
weight_kg=[50,53,56,59,62,66,68,72]
age_year=[20,33,25,28,30,35,38,40]
```

```python
plt.figure(figsize=(8,5))
```

```python
plt.scatter(height_cm,weight_kg,c=age_year, cmap='rainbow', s=120, edgecolor='black')
plt.colorbar(label='Age (Years)')
plt.title('Scatter Plot: Height Vs Weight (Colored by age)')
plt.xlabel('Height (cm)')
plt.ylabel('Weight (kg)')
plt.grid(color='blue', linestyle='--', linewidth=0.7)
plt.show()
```

# Pie Plot Example
 ## plt.pie(x,label='', colors='',autopct='',startangle='',shadow=,explode=,radius=,..and so on ,optional)

```python
depts=['HR','WEB-DEVELOPMENT','APP-DEVELOPEMENT','AGILE-SCRUM','TESTING']
expenses=[12000,25000,18000,15000,20000]
```

```python
plt.figure(figsize=(7,7))
```

```python
plt.pie(expenses,labels=depts,autopct='%1.2f%%',startangle=140,explode=(0,0.05,0,0,0),shadow=True)
plt.title('Department wise Expenses Report')
plt.show()
```

```python
plt.pie(expenses,labels=depts,colors=plt.cm.plasma([0.6,0.3,0.5,0.7,0.9]),autopct='%1.2f%%',startangle=140,explode=(0,0.05,0,0,0),shadow=True)
plt.title('Department wise Expenses Report')
plt.show()
```

# Create a pie plot usng following data
## sales_data={'Mobile':45000,'Clothing':30000,'Groceries':35000,'Home Decor':20000}
### show percentage(%) with two decimal
### use colormap to give the difference color to your pie

```python
sales_data = {'Mobile': 45000, 'Clothing': 30000, 'Groceries': 35000, 'Home Decor': 20000}
```

```python
category = list(sales_data.keys())
sales = list(sales_data.values())
```

```python
plt.figure(figsize=(6.4,4.8))
```

```python
plt.pie(sales,labels=category,colors=plt.cm.summer([0.6,0.3,0.5,0.7,0.9]),textprops={'color':'black','fontsize':10,'fontweight':'bold'} ,autopct='%1.2f%%', startangle=140, shadow=True)
plt.title('Sales Report by Category')
plt.savefig('categorieswisesale.png',dpi=300,bbox_inches='tight') #plt.savefig('filenamewithextension',dpi=value,bbox_inches='')
plt.show()
```

```python
fig = plt.gcf()       # Get current figure (size chart)
print(fig.get_size_inches())
```

# Histogram Example

```python
marks=[10,20,30,35,45,55,78,90,34,12,80,90,45,100,0,50,50,12,56,90,80,11,7]
```

```python
plt.style.use('ggplot')
plt.figure(figsize=(7,4))
plt.hist(marks,bins=5,color='green',edgecolor='darkblue',alpha=0.8)
plt.title('Histogram for Student Marks')
plt.xlabel('Marks Range')
plt.ylabel('Number of Students')
plt.grid(color='blue', linestyle='--', linewidth=0.7)
plt.show()
```

```python
plt.style.use('ggplot')
plt.figure(figsize=(7,4))

n, bins, patches = plt.hist(marks, bins=5, edgecolor='darkblue', alpha=0.8)
bar_colors = ['red', 'blue', 'green', 'yellow', 'orange']

for patch, color in zip(patches, bar_colors):patch.set_facecolor(color)

plt.title('Histogram for Student Marks')
plt.xlabel('Marks Range')
plt.ylabel('Number of Students')
plt.grid(color='blue', linestyle='--', linewidth=0.7)
plt.show()
```
