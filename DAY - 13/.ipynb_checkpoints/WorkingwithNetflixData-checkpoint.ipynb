```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('netflix_data.csv')
print(df.head())
```

```python
df=df.dropna(subset=['release_year'])
```

```python
df['release_year']=df['release_year'].astype(int)
```

```python
plt.figure(figsize=(6,4))
plt.hist(df['release_year'],bins=100,color='green',edgecolor='darkblue',alpha=0.8)
plt.title('Netflix released Titles by Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.grid(color='blue', linestyle='--', linewidth=0.7)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv('netflix_data.csv')
df=df.dropna(subset=['release_year'])
df['release_year']=df['release_year'].astype(int)

plt.figure(figsize=(6,4))
plt.hist(df['release_year'],bins=100,color='green',edgecolor='yellow',alpha=0.8)
plt.title('Netflix released Titles by Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.grid(color='blue', linestyle='--', linewidth=0.7)
plt.tight_layout()
plt.savefig('netflix_bar.pdf',dpi=300,bbox_inches='tight') #untuk save pdf,png & etc file
plt.show()
```

```python
df=df.dropna(subset=['release_year','type'])
```

```python
df['release_year']=df['release_year'].astype(int)
```

```python
movies=df[df['type']=='Movie']
shows=df[df['type']=='TV Show']
```

```python
plt.figure(figsize=(6,4))
plt.hist([movies['release_year'],shows['release_year']],
         bins=30,
         color=['green','red'],
         edgecolor='yellow',
         alpha=0.8,
         label=['Movies','TV-Shows'])

plt.title('Netflix released by Year & Type')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles Released')
plt.grid(color='blue', linestyle='--', linewidth=0.7)
plt.tight_layout()
plt.legend(title='Movies/Shows')
plt.savefig('netflix_showsmovies.jpg',dpi=300,bbox_inches='tight') #untuk save pdf,png & etc file
plt.show()
```

# Use netflix_data.csv dataset
### print how many Movies and TV Shows are available on netflix
### Display Movies and TV Shows proportion in a pie plot
### Save file as pdf

```python
df = pd.read_csv('netflix_data.csv')
```

```python
type_count = df['type'].value_counts()
print("Movies and TV Shows count:")
print(type_count)
```

```python
plt.figure(figsize=(6,5))
colors = ['skyblue', 'lightcoral']
plt.pie(type_count,
        labels=type_count.index,
        autopct='%1.2f%%',
        startangle=45,
        colors=colors,
        shadow=True,
        explode=(0.05,0.05),
        textprops={'color':'black','fontsize':10})

plt.title('Netflix Content Type Proportion', fontsize=14)
plt.tight_layout()
plt.savefig('netflix_type_Proportion_pieplot.pdf', dpi=300, bbox_inches='tight')
plt.show()
```
