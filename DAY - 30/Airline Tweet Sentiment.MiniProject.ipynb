**Steps:**

6. Train Logistic Regression Model.
7. Evaluate accuracy and classification report.

```python
# Step 1a: Import Required Libraries
import pandas as pd
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Step 1b: Download nltk required thing like stopwords & etc
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()
```

```python
# Step 2a: Load DataSet
df = pd.read_csv('airline_tweets_sample.csv')
```

```python
# Step 2b: Check DataSet at least top 5 values
print(df.head(5))
```

```python
# Step 3: Text Cleaning and Preprocessing
# For each tweet do:
# Convert to lowercase
# Removing URLs
# Remove special characters and numbers
# Remove stopwords (common words like *the,is,and* etc.)
# Apply **stemming** (reduce words to their root form)

def preprocess_text(text):

    text = text.lower()
    text = re.sub(r'http\S+|www.\S+', '', text)
    text = re.sub(r'@\w+|#\w+', '', text)
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    words = text.split()


    words = [word for word in words if word not in stop_words]
    words = [stemmer.stem(word) for word in words]
    return ' '.join(words)

df['cleaned_text'] = df['text'].apply(preprocess_text)
print("Preprocessing complete.")
```

```python
# Step 4:
# Convert text to numerical vectors (TF-IDF)
# Check X,y and shape len

vectorizer = TfidfVectorizer(max_features=5000)
X = vectorizer.fit_transform(df['cleaned_text'])
y = df['sentiment']

print(f"Shape of feature matrix X: {X.shape}")
print(f"Length of target vector y: {len(y)}")
```

```python
# Step 5: Split into train and test sets
# 80% training and 20% testing
X_train, X_test, y_train, y_test = train_test_split(X, y,
                                                    test_size=0.20,
                                                    random_state=42,
                                                    stratify=y)
```

```python
# Step 6: Train a Logistic Regression model
# Create Logistic Model
# Train Logistic Model

model = LogisticRegression(max_iter=1000, random_state=42)
```

```python
model.fit(X_train, y_train)
```

```python
# Step 7: Evaluate accuracy and classification report
# Predict model
# Predcision, recall, F1-Score for each sentiment

y_pred = model.predict(X_test)
```

```python
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy on Test Set: {accuracy:.4f}")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
# Step 8: Predict sentiment for new example tweets

new_tweets = [
    "This flight was cancelled, service was awful.",
    "The crew was incredibly polite and the snacks were great. A perfect flight!",
    "My flight was delayed, but it was an average experience.",
]

cleaned_new_tweets = [preprocess_text(tweet) for tweet in new_tweets]
X_new = vectorizer.transform(cleaned_new_tweets)
predictions = model.predict(X_new)

for i, (tweet, prediction) in enumerate(zip(new_tweets, predictions)):
    print(f"\nTweet {i+1}: '{tweet}'")
    print(f"--> Predicted Sentiment: {prediction.upper()}")
```
