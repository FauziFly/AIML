# Searching and Sorting Example in 1D Array

```python
sorted_arr=np.sort(our_arr)
print('***Sorted Array***')
print(sorted_arr)
```

```python
print('Is there any item greater than 500?: ',np.any(sorted_arr>500))
```

```python
print('All greater than 99?: ',np.all(sorted_arr>99))
```

```python
print('Where values >500',np.where(sorted_arr>500))
```

```python
print('All Number greater than 500')
for num in np.where(sorted_arr>500):
    print (num)
```

```python
print('All Number greater than 500')
for num in sorted_arr>500:
    print (num)
```

```python
selected_values=sorted_arr[sorted_arr>500]
```

```python
print(selected_values)
```

# Slicing 1D Array
    1. To take elements from starting index to end index
    2. Example - [start:end]

```python
print(sorted_arr)
```

```python
slice_our_arr=sorted_arr[1:3]
print(slice_our_arr)
```

# Starting from an index and going till last

```python
print(sorted_arr[3:])  #[3:] means → start from index 3 (the 4th element) and take everything until the end
```

```python
print(sorted_arr[:4])  # [:4] means → start from the beginning (index 0) and stop before index 4
```

```python
print(sorted_arr[:5])
```

# Step
    ## [start:end:step]
    ## Step: every step element

```python
print(sorted_arr)
```

```python
our_slice=sorted_arr[1:6]
print(our_slice)
```

```python
every_third_element=sorted_arr[::3]  #[start : stop : step] - start → empty, so it starts from the beginning, stop → empty, so it goes until the end, step = 3 → take every 3rd element
print(every_third_element)
```

```python
print('Original Array')
print(sorted_arr)
print('Every Third Element')
every_third_element=sorted_arr[::3]
print(every_third_element)
```

# Slice Reverse

```python
print('Original Array')
print(sorted_arr)
print('Revers Array: ')
print(sorted_arr[::-1])
```

```python
print('Last five elements')
print(sorted_arr[-5:])
```

```python
print('Slicing Done')
```
