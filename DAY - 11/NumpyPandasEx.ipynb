```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_excel('students.xlsx')
print(df)
```

```python
marks=df['Marks'].to_numpy()
print('Numpy Array of Marks')
print(marks)
```

```python
print('Highest Mark',np.max(marks))
```

```python
print('Lowest Mark',np.min(marks))
```

```python
print('Average Mark',np.mean(marks))
```

# .loc:label based
# .iloc: index based

```python
print('Original Dataframe') # pandas method
print(df)
print('Location: Label Based')
print(df.loc[2,['Name','Marks']])  
```

```python
print('Original Dataframe')  # pandas method
print(df)
print('Location: Index Based')
print(df.iloc[2,[1,2]]) 
```

# Axis in Numpy: 
    1. Axis 0: vertical direction: works columns-wise
    2. Axis 1: horizontal direction: works rows-wise

```python
print('Sum down in each column',np.sum(marks_arr,axis=0))
```

```python
print('Sum down in each row',np.sum(marks_arr,axis=1))
```

# Find the student name who got highest marks

```python
topper=df.loc[df['Marks'].idxmax(),'Name']
print('Top Student Name',topper)
```

```python
last=df.loc[df['Marks'].idxmin(),'Name']
print('Last Student Name',last)
```

```python
first_in_class = df.loc[df['Marks'].idxmax(), 'Name']
print('First in Class Student Name:', first_in_class)
```
