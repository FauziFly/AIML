# We are going to write one Multi Label Classification Example using TnesorFlow here

```python
import tensorflow as tf
import pandas as pd
from sklearn.preprocessing import MultiLabelBinarizer
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
data={
    'overview':
    [
        'A Hero Saves the city from destruction',
        'A girl and boy become friend during war',
        'A funny story of collegues during office meet',
        'One scary nights in a deserted place'
    ],
    'genre':[['Action','Adventure'],['Romance','Drama'],['Comedy'],['Horror']]
}
```

```python
df=pd.DataFrame(data)
df
```

```python
mlb=MultiLabelBinarizer()
y=mlb.fit_transform(df['genre'])
```

```python
y
```

```python
tokenizer=Tokenizer(num_words=10000,oov_token='<OOV>')
tokenizer.fit_on_texts(df['overview'] )
X=tokenizer.texts_to_sequences(df['overview'])
X=pad_sequences(X,maxlen=20,padding='post')
```

```python
# model creation

model=tf.keras.Sequential([
    tf.keras.layers.Embedding(input_dim=5000,output_dim=16,input_length=20),
    tf.keras.layers.GlobalAveragePooling1D(),
    tf.keras.layers.Dense(16,activation='relu'),
    tf.keras.layers.Dense(len(mlb.classes_),activation='sigmoid'),
])

```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
model.fit(X,y,epochs=20)
```

```python
history=model.fit(X,y,epochs=20)
```

```python
# Predict on a new movie overview
new_movie_overview= ['A lonely wolf was walking in the forest']
seq=tokenizer.texts_to_sequences(new_movie_overview)
seq=pad_sequences(seq,maxlen=20,padding='post')
prediction=model.predict(seq)
prediction_labels=[mlb.classes_[i] for i in range(len(prediction[0])) if prediction[0][i]>0.5]
print('Predicted Genres:',prediction_labels)
```

```python
# use this dataset to write one more MultiLabel Classification prediction using Neural Network
data = {
    "headline": [
        "Government launches new digital economy policy",
        "Star player wins championship after dramatic final",
        "Tech company releases AI-powered smartphone",
        "Stock market drops amid political tensions",
        "New environmental policy gains global support",
        "Space agency announces manned Mars mission"
    ],
    "topics": [
        ["Politics", "Economy", "Technology"],
        ["Sports"],
        ["Technology", "Business"],
        ["Economy", "Politics"],
        ["Environment", "Politics"],
        ["Science", "Technology"]
    ]
}
```

```python
df2 = pd.DataFrame(data)
```

```python
mlb2 = MultiLabelBinarizer()
y2 = mlb2.fit_transform(df2['topics'])
```

```python
tokenizer2 = Tokenizer(num_words=10000, oov_token='<OOV>')
tokenizer2.fit_on_texts(df2['headline'])
X2 = tokenizer2.texts_to_sequences(df2['headline'])
X2 = pad_sequences(X2, maxlen=20, padding='post')
```

```python
model2 = tf.keras.Sequential([
    tf.keras.layers.Embedding(input_dim=len(tokenizer2.word_index) + 1, output_dim=16, input_length=20),
    tf.keras.layers.GlobalAveragePooling1D(),
    tf.keras.layers.Dense(16, activation='relu'),
    tf.keras.layers.Dense(len(mlb2.classes_), activation='sigmoid')
])
```

```python
model2.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
model2.fit(X2, y2, epochs=50)
```

```python
new_headline = ['New policy on space exploration and technology announced']
seq2 = tokenizer2.texts_to_sequences(new_headline)
seq2 = pad_sequences(seq2, maxlen=20, padding='post')
prediction2 = model2.predict(seq2)
```

```python
prediction_labels2 = [mlb2.classes_[i] for i in range(len(prediction2[0])) if prediction2[0][i] > 0.5]
print('Predicted Topics:', prediction_labels2)
```
