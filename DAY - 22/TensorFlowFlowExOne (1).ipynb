**Arithmetic Operation with Tensors**


*  Add
*  Sub
*  Multi
*  Div

```python
tensor_one=tf.constant([[11,9],[12,5]])
tensor_one
```

```python
#tensor_one+15
tf.add(tensor_one,15)
```

```python
#tensor_one-5
tf.subtract(tensor_one,5)
```

```python
#tensor_one*2
tf.multiply(tensor_one,2)
```

```python
#tensor_one/2
tf.divide(tensor_one,2)
```

```python
tf.square(tensor_one)
```

**Exercise**

Create two tensors with random values between 0 and 1, shape is [6,300]
Find out the follow details:

* Number of elements
* Rank,
* Shape

```python
tensor_a = tf.random.uniform(shape=[6, 300], minval=0, maxval=1)
tensor_b = tf.random.uniform(shape=[6, 300], minval=0, maxval=1)
```

```python
num_elements_a = tf.size(tensor_a)
rank_a = tf.rank(tensor_a)
shape_a = tf.shape(tensor_a)

num_elements_b = tf.size(tensor_b)
rank_b = tf.rank(tensor_b)
shape_b = tf.shape(tensor_b)
```

```python
print("=== Tensor A ===")
print("Shape:", shape_a.numpy())
print("Rank:", rank_a.numpy())
print("Number of elements:", num_elements_a.numpy())

print("\n=== Tensor B ===")
print("Shape:", shape_b.numpy())
print("Rank:", rank_b.numpy())
print("Number of elements:", num_elements_b.numpy())
```
