```python
# Import all required libraries
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import matplotlib.pyplot as plt
import seaborn as sns
from numpy import vectorize
```

```python
df=pd.read_csv('/content/UpdatedResumeDataSet.csv',encoding='utf-8')
print('Shape of data frame',df.shape)
df.head(10)
```

```python
print('\n Columns',df.columns.tolist())
```

```python
sample_df=df.sample(10,random_state=42).reset_index(drop=True)
sample_df
```

```python
sample_df=sample_df.rename(columns={'Resume':'Resume_Text'})
sample_df
print(sample_df.shape)
```

```python
vectorizer=TfidfVectorizer(stop_words='english')
tfidf_matrix=vectorizer.fit_transform(sample_df['Resume_Text'])
tfidf_matrix.shape
```

```python
similartiness_matrix=cosine_similarity(tfidf_matrix)
sim_df=pd.DataFrame(similartiness_matrix,index=sample_df['Category']+'_'+sample_df.index.astype(str),
                    columns=sample_df['Category']+'_'+sample_df.index.astype(str))

sim_df.round(2).head(3)
```

```python
resumes={
    'candidates':['Sam','Amit','Maan'],
    'resume_text':['Python Developer Skill in Django,Flask,Rest API and MySQL',
                   'Machine Learning engineer experience in Python,TenseFlow,NLP and data analysis',
                   'Full stack developer with .net, javascript,React,Noda.js and MongoDB experience']
}
```

```python
df=pd.DataFrame(resumes)
df
```

```python
# Tambah candidate lagi
df.loc[len(df)]=['Sav','Backend developer skilled in Python, Flask and SQL']
```

```python
vectorize=TfidfVectorizer(stop_words='english')
tfidf_matrix=vectorizer.fit_transform(df['resume_text'])
tfidf_matrix.shape
tfidf_matrix
```

```python
similartiness_matrix=cosine_similarity(tfidf_matrix)
sim_df=pd.DataFrame(similartiness_matrix,index=df['candidates'],columns=df['candidates'])
sim_df.round(2)
```
