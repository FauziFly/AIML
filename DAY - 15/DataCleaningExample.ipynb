```python
import pandas as pd
```

```python
data={'Students':['Amit','Anam','Chi','Sam','Ravi','Jai'],
 'Marks':[85,None,65,75,None,55]}
```

```python
df=pd.DataFrame(data)
print('Original Data')
df
```

```python
print('Missing Data')
print('Number of Missing values are:',df.isnull().sum())
```

```python
average_marks=df['Marks'].mean()
print(average_marks)
```

```python
df.fillna(average_marks, inplace=True)
```

```python
print('after fillna() call')
df
```

# Download students_performance_dirty.csv
## find out number of missing data
## drop rows where gender is missing
## fill study_hours with average_hours

```python
# 1. Read dataset
df = pd.read_csv('students_performance_dirty.csv')
```

```python
# 2. Show first few rows (optional)
print(df.head())
```

```python
# 3. Find number of missing data
print("\nNumber of missing values in each column:")
print(df.isnull().sum())
```

```python
# 4. Drop rows where 'gender' is missing
df=df.dropna(subset=['gender'])
```

```python
# 5. Fill 'study_hours' with the average value
df['study_hours'] = df['study_hours'].fillna(df['study_hours'].mean())
```

```python
# 6. Verify changes
print("\nAfter cleaning:")
print(df.isnull().sum())
```

```python
df = pd.read_csv('students_performance_dirty.csv')
print(df.head())

print("\nNumber of missing values in each column:")
print(df.isnull().sum())

df=df.dropna(subset=['gender'])
df['study_hours'] = df['study_hours'].fillna(df['study_hours'].mean())

print("\nAfter cleaning:")
print(df.isnull().sum())
```
