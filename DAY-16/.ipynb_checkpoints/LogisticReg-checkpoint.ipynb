# 2. Logistic Regression - Pass/Fail

```python
# imported all required libraries
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
```

```python
# collect data
df=pd.read_csv('pass_fail.csv')
df.head()
```

```python
# split the data
X= df[['Hours']]
y= df['Result']
```

```python
# create the model(change here only, we are using Logistic Regression)
model=LogisticRegression()
```

```python
# trained the model
model.fit(X,y)
```

```python
pred_35= model.predict([[3.5]])
print('Predicted result for 3.5 hours:', pred_35[0])
```

```python
hours_study=int(input('Number of hours studied'))
pred_value=model.predict([[hours_study]])
print(f'After {hours_study} you are likely to {pred_value[0]}')
```

```python
# imported all required libraries
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import LabelEncoder
```

```python
df=pd.read_csv('customer_purchase.csv')
df.head()
```

```python
# data split and labelling
le= LabelEncoder()
df['Purchased'] = le.fit_transform(df['Purchased'])
X = df[['Age', 'Salary']]
y= df['Purchased']
```

```python
df
```

```python
# creating the model
model=LogisticRegression()
```

```python
# training the  model
model.fit(X,y)
```

```python
# Prediction
# assign fix value
new_cust = [[28,35000]]
prediction = model.predict(new_cust)
print(f'Predicted result (1=Yes, 0=No): {prediction[0]}')
```

```python
# Prediction: Predict the result
new_cust_age = int(input('Enter Age'))
new_cust_sal = int(input('Enter Salary'))
new_cust = [[new_cust_age,new_cust_sal]]
new_cust = model.predict(new_cust)
if (prediction):
    print('Likely Purchased')
else:
    print('Likely Not Purchased')
```
