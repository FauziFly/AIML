```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import numpy as np
```

# 1. Linear regression - house of price
## download house_price.csv file from github
## **Goal:** Train a Linear Rgeression model to predict hosue price from Size_sqft.


```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
df=pd.read_csv('house_price.csv')
df.head()
```

```python
df = df.replace('?', np.nan)
df = df.dropna()
df['Size_sqft'] = df['Size_sqft'].astype(float)
df['Price'] = df['Price'].astype(float)
```

```python
X = df[['Size_sqft']]
y = df['Price']
```

```python
model = LinearRegression()
model.fit(X, y)
print("Training done successfully!")
```

```python
print("Coefficient (Slope):", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
new_size = float(input("Enter house size (sqft) to predict price: "))
new_data = pd.DataFrame({'Size_sqft': [new_size]}) 
predicted_price = model.predict(new_data)
print(f"Predicted price for {new_size} sqft house is: ${predicted_price[0]:.2f}")
```
